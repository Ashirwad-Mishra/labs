```python
import matplotlib.pyplot as plt

# Function to generate Fibonacci numbers
def fibonacci(n):
    fib_sequence = [0, 1]
    for i in range(2, n):
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
    return fib_sequence

# Generate the first 100 Fibonacci numbers
fib_numbers = fibonacci(100)

# Plotting the Fibonacci numbers
plt.figure(figsize=(10, 5))
plt.plot(fib_numbers, marker='o')
plt.title('First 100 Fibonacci Numbers')
plt.xlabel('Index')
plt.ylabel('Fibonacci Number')
plt.grid()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Define the x values within the valid range
x = np.linspace(-1, 1, 100)

# Define the function y = sqrt(1 - x^2)
y = np.sqrt(1 - x**2)

# Create the plot
plt.plot(x, y)

# Add labels and title to the plot
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Plot of y = sqrt(1 - x^2)")

# Display the plot
plt.show()
```
