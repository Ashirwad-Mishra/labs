```python
data = [
    [ "Meat", 250, 40,8],
    ["Banana" , 130, 58, 5],
    ["Avocado" , 140, 20, 3],
    ["Sweet" , 120, 30 , 6]]
```

```python
pd.DataFrame(data , columns = ["Food" , "Calories" , "Potassium" , "Fat"] , index = [1,2,3,4])
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dataset
foods = [
    "Meat", "Banana", "Avocados", "Sweet Potatoes", "Spinach",
    "Watermelon", "Coconut water", "Beans", "Legumes", "Tomato"
]
calories = [250, 130, 140, 120, 20, 20, 10, 50, 40, 19]
potassium = [40, 55, 20, 30, 40, 32, 10, 26, 25, 20]
fat = [8, 5, 3, 6, 1, 1.5, 0, 2, 1.5, 2.5]

x = np.arange(len(foods))  # Bar positions

# Plotting
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.25

ax.bar(x - bar_width, calories, bar_width, label="Calories", color="blue")
ax.bar(x, potassium, bar_width, label="Potassium", color="green")
ax.bar(x + bar_width, fat, bar_width, label="Fat", color="orange")

# Labeling
ax.set_xticks(x)
ax.set_xticklabels(foods, rotation=45, ha="right")
ax.set_xlabel("Food Items")
ax.set_ylabel("Content")
ax.set_title("Comparison of Food Nutritional Content")
ax.legend()

plt.tight_layout()
plt.show()
```
