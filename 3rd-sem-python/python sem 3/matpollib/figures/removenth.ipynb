```typescript
def printTriangle(n):
    for i in range(1, n+1):
        for j in range(i):
            print("*", end = " ")
        print()
```

```typescript
printTriangle(5)
```

```typescript
printTriangle(10)
```

```typescript
printTriangle(15)
```
