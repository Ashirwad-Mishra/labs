```python
import matplotlib.pyplot as plt

center = (0, 3)
radius = 7

fig, ax = plt.subplots()

circle = plt.Circle(center, radius, edgecolor='blue', facecolor='none')

ax.add_patch(circle)

ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)

plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Circle with Center and Radius")

plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
